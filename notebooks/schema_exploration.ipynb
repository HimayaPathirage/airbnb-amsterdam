```python
import pandas as pd
print(pd.__version__)
```

```python
##----------------------Listings.csv.gz-----------------
df_listings = pd.read_csv('../data/raw/listings.csv.gz', compression='gzip')
print(df_listings.shape) ## Number of rows and columns
print(df_listings.columns.tolist()) ## Names of all the columns
print(df_listings.dtypes) ## Data types for each columns
```

```python
for col in df_listings.columns:
    print(col)
```

```python
null_pct = (df_listings.isnull().mean() * 100).round(1).sort_values(ascending=False)
print(null_pct)
```

```python
##-----------------------------calender.csv.gz--------------------------------------
df_calendar = pd.read_csv('../data/raw/calendar.csv.gz', compression='gzip', nrows=5)
print(df_calendar.shape)
print(df_calendar.columns.tolist())
print(df_calendar.dtypes)
print(df_calendar.head())
```

```python
df_calendar_sample = pd.read_csv('../data/raw/calendar.csv.gz', compression='gzip', nrows=2000)
print(df_calendar_sample['available'].value_counts())
print(df_calendar_sample.groupby('available')['price'].apply(lambda x: x.isnull().mean() * 100))
```

```python
df_calendar_full = pd.read_csv('../data/raw/calendar.csv.gz', compression='gzip', usecols=['listing_id', 'available', 'price'])
print(df_calendar_full.shape)
print(df_calendar_full['price'].isnull().mean() * 100)
print(df_calendar_full['available'].value_counts())
```
