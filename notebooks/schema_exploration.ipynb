```python
price_numeric = df_listings['price'].str.replace('$', '', regex=False).str.replace(',', '', regex=False).astype(float)

Q1 = price_numeric.quantile(0.25)
Q3 = price_numeric.quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print(f"Q1: {Q1}, Q3: {Q3}, IQR: {IQR}")
print(f"Lower bound: {lower_bound}, Upper bound: {upper_bound}")
print(f"Outliers below: {(price_numeric < lower_bound).sum()}")
print(f"Outliers above: {(price_numeric > upper_bound).sum()}")
print(price_numeric.describe())
```

```python
top_prices = price_numeric.sort_values(ascending=False).head(15)
print(top_prices)

top_listings = df_listings.loc[top_prices.index, ['id', 'name', 'price', 'room_type', 'accommodates']]
print(top_listings)
```
